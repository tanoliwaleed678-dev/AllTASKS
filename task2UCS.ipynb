```python
import heapq

def uniform_cost_search(graph, start_node, goal_node):
    # priority_queue stores tuples: (cumulative_cost, current_node, path_list)
    # The heapq module will always pop the tuple with the lowest cumulative_cost
    priority_queue = [(0, start_node, [start_node])]

    # Keep track of the minimum cost to reach each node
    visited = {}

    while priority_queue:
        # Pop the lowest cost path
        cost, current, path = heapq.heappop(priority_queue)

        # Goal Test: The first time we pop the goal, it's the optimal path
        if current == goal_node:
            return path, cost

        # Skip if we've already found a cheaper way to this node
        if current in visited and visited[current] <= cost:
            continue

        visited[current] = cost

        # Explore neighbors
        for neighbor, weight in graph.get(current, {}).items():
            new_cost = cost + weight
            new_path = path + [neighbor]
            heapq.heappush(priority_queue, (new_cost, neighbor, new_path))

    return None, float('inf')

# Graph structure based on the weights in your image
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 1, 'E': 3},
    'C': {'F': 5},
    'D': {'G': 2},
    'E': {'D': 1},
    'F': {'G': 2},
    'G': {}
}

# Run the search
path, total_cost = uniform_cost_search(graph, 'A', 'G')

# Output results
if path:
    print(f"Optimal Path: {' -> '.join(path)}")
    print(f"Total Minimum Cost: {total_cost}")
else:
    print("No path found.")
```
