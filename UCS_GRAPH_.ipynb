```python
import heapq

def uniform_cost_search(graph, start, goal):
    # Priority Queue stores (cost, current_node, path)
    # Using a heap ensures we always pop the node with the minimum cost
    priority_queue = [(0, start, [start])]

    # Keep track of the minimum cost to reach each node
    visited = {}

    while priority_queue:
        (cost, current_node, path) = heapq.heappop(priority_queue)

        # Goal Check
        if current_node == goal:
            return path, cost

        # If we've found a cheaper way to this node already, skip it
        if current_node in visited and visited[current_node] <= cost:
            continue

        visited[current_node] = cost

        # Explore neighbors
        for neighbor, weight in graph.get(current_node, {}).items():
            new_cost = cost + weight
            if neighbor not in visited or new_cost < visited[neighbor]:
                heapq.heappush(priority_queue, (new_cost, neighbor, path + [neighbor]))

    return None, float('inf')

# --- Example Usage ---
# Graph represented as an adjacency list: { node: { neighbor: cost } }
graph = {
    'A': {'B': 2, 'C': 5},
    'B': {'D': 2, 'E': 4},
    'C': {'G': 10},
    'D': {'G': 5},
    'E': {'G': 1},
    'G': {}
}

path, total_cost = uniform_cost_search(graph, 'A', 'G')

print(f"Optimal Path: {path}")
print(f"Total Cost: {total_cost}")
```
